# Run time analysis - varying dosing grid half length

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
dose_grid = np.array([0.5, 1.0, 1.5, 2.0, 2.5, 3.0])
run_time = np.array([3396.24, 3560.97, 3217.57, 3091.06, 2992.11, 2953.64])

fig, ax = plt.subplots()
line_format = "o"
ax.plot(dose_grid, run_time/3600, line_format)
ax.set_title("Run time vs half length, num_Hx = 1e7, num_bins = 51")
ax.set_ylabel("Runtime (hours)")
ax.set_xlabel("Dose grid half length")
```
