# Run time analysis: 1e7 / 2.5 cm / 10 - 105 bins

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
run_time = np.array([595.604, 882.944, 2989.48, 9599.48, 13006.1, 18580, 24025.90, 25283])
num_bins = np.array([10, 26, 51, 76, 85, 95, 101, 105])
```

```python
# Get a simple fit of the data
def runtime_exp(x, b):
    return np.exp(b*x)

popt, pcov = curve_fit(runtime_exp, num_bins, run_time/3600)

print(np.linalg.cond(pcov))
print(popt)
print(np.diag(pcov))
```

```python
fig, ax = plt.subplots()
line_format = "o"
ax.plot(num_bins, run_time/3600, line_format)
ax.plot(num_bins, runtime_exp(num_bins, *popt), color="red")

ax.set_title("Run time vs bins, num_Hx = 1e7, HL = 2.5 cm")
ax.set_xlabel("Number bins")
ax.set_ylabel("Computation time (hours)")
```

```python
# Predict time to simulate with # bins using fit from above
bins = 180

predicted_time = runtime_exp(bins, *popt)
print(predicted_time)
```
