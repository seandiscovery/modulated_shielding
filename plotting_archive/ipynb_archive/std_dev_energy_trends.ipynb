# Standard deviation trends
This notebook explores how the standard deviation per voxel changes as a function of history number, radius, and angle.

```python
from mosh_src.plotting_functions import *
from math import pi
import matplotlib.pyplot as plt
import matplotlib.colors as colors
from matplotlib import ticker
from decimal import Decimal
```

The block below assumes a cubic dose space (i.e. XBins = YBins = ZBins)

```python
# Define parameters and import TOPAS files
files = np.array(["../Outputs_0724/bin_0.5mm_hist_10e7.csv", "../Outputs_0724/bin_1mm_hist_10e7.csv", 
	"../Outputs_0724/bin_0.5mm_hist_10e6.csv", "../Outputs_0724/bin_1mm_hist_10e6.csv"])

angle = pi
radius = 0.3
num_files = np.shape(files)[0]

num_Hx = np.array([1e7, 1e7, 1e6, 1e6])
num_bins = np.array([101, 51, 101, 51])
dose_grid_half = np.array([2.55, 2.55, 2.55, 2.55])
plot_colors = ["blue", "red", "green", "yellow", "orange"]

# Use source_transZ to plot in the z plane of the source
# Use z_slices to plot an arbitrary z plane for each file
source_transZ = np.array([-0.5, -0.5, -0.5, -0.5])
z_slices = np.array([25, 25, 25, 25])
```

```python
# Extract energy along a constant angle and constant radius for each file
energy_storage = []
for i in range(num_files):
    source_plane = cm_to_voxel(source_transZ[i], num_bins[i], dose_grid_half[i]*2/num_bins[i])
        
    # If plotting an arbitrary z plane, use "z_slices[i]" as fifth argument
    # If plotting plane of source, use "source_plane" as fifth argument
    eng_at_const_theta, eng_at_const_r = energy_along_theta(files[i], angle, (num_bins[i], num_bins[i]), 
                                                            dose_grid_half[i], source_plane, err=0.01, radius=radius)

    energy_storage.append((eng_at_const_theta, eng_at_const_r))
```

### Step 1: plot std dev as function of radius

```python
# Extract parameters to plot = independent variable (radius), dependent variable (energy), standard deviation
# Standard deviation used for error bars and to plot std dev vs radius below
plot_storage_1 = []
plot_storage_2 = []
plot_legend = []
plot_range_1 = (0.3, 1.0)
for j in range(num_files):    
    """
    To plot std dev as function of radius:
        - eng_to_plot = energy_storage[j][0]
        - ind_var = 0
        - Adjust plot_rng as desired (entire range --> plt_rng = (0,0))
    
    To plot std dev as function of theta:
        - eng_to_plot = energy_storage[j][1]
        - ind_var = 1
        - Adjust plot_rng as desired
    
    To save figures of energy vs radius/theta, set save=True
    Don't forget to adjust the plot title when changing the independent variable!
    """
    eng_to_plot = energy_storage[j][0]
    # Plot restricted range (0.3, 1.0)
    plot_tuple_1 = plotter_1D(eng_to_plot, ind_var=0, plot_rng=plot_range_1, std_dev=True, num_Hx=num_Hx[j])
    # Plot full range
    plot_tuple_2 = plotter_1D(eng_to_plot, ind_var=0, std_dev=True, num_Hx=num_Hx[j])
    
    plot_storage_1.append(plot_tuple_1)
    plot_storage_2.append(plot_tuple_2)
    plot_legend.append(f"{'%.2E' % Decimal(num_Hx[j])}_{num_bins[j]}_{dose_grid_half[j]}")
```

```python
# Plot std dev of sum vs radius at constant angle (theta) for range = (0.3, 1.0)
fig1, ax1 = plt.subplots()

line_handles = []
line_format = "o" # set to "o" for scatter plot
for k in range(len(plot_storage_1)):	
	line, = ax1.plot(plot_storage_1[k][0], plot_storage_1[k][2], line_format, color=plot_colors[k], label=plot_legend[k])
	line_handles.append(line)

ax1.legend(handles=line_handles)
ax1.set_title("Std dev of sum vs radius, theta = pi")
```

```python
# Plot std dev of sum vs radius at constant angle for entire range
fig3, ax3 = plt.subplots()

line_handles = []
line_format = "o" # set to "o" for scatter plot
for k in range(len(plot_storage_2)):	
    line, = ax3.plot(plot_storage_2[k][0], plot_storage_2[k][2], line_format, color=plot_colors[k], label=plot_legend[k])
    line_handles.append(line)

ax3.legend(handles=line_handles)
ax3.set_title("Std dev of sum vs radius, theta = pi")
```

```python
# Plot std dev of mean vs radius for range (0.3, 1.0)
fig4, ax4 = plt.subplots()

line_handles = []
line_format = "o" # set to "o" for scatter plot
for k in range(len(plot_storage_1)):	
    line, = ax4.plot(plot_storage_1[k][0], plot_storage_1[k][3], line_format, color=plot_colors[k], label=plot_legend[k])
    line_handles.append(line)

ax4.legend(handles=line_handles)
ax4.set_title("Std dev of mean vs radius, theta = pi")
```

```python
# Plot std dev of mean vs radius for entire range
fig5, ax5 = plt.subplots()

line_handles = []
line_format = "o" # set to "o" for scatter plot
for k in range(len(plot_storage_2)):	
    line, = ax5.plot(plot_storage_2[k][0], plot_storage_2[k][3], line_format, color=plot_colors[k], label=plot_legend[k])
    line_handles.append(line)

ax5.legend(handles=line_handles)
ax5.set_title("Std dev of mean vs radius, theta = pi")
```

### Step 2: plot std dev as a function of angle

```python
# Note that we reset variables in this block
# Extract parameters to plot = independent variable (radius), dependent variable (energy), standard deviation
plot_storage = []
for j in range(num_files):
    """
    To plot std dev as function of radius:
        - eng_to_plot = energy_storage[j][0]
        - ind_var = 0
        - Adjust plot_rng as desired (entire range --> plt_rng = (0,0))
    
    To plot std dev as function of theta:
        - eng_to_plot = energy_storage[j][1]
        - ind_var = 1
        - Adjust plot_rng as desired
    
    To save figures of energy vs radius/theta, set save=True
    Don't forget to adjust the plot title when changing the independent variable!
    """
    eng_to_plot = energy_storage[j][1]
    plot_tuple = plotter_1D(eng_to_plot, ind_var=1, plot_rng=(0, pi), std_dev=True, num_Hx=num_Hx[j])
    
    plot_storage.append(plot_tuple)
```

```python
# Plot std dev of sum as function of angle at constant radius
fig2, ax2 = plt.subplots()

line_handles = []
line_format = "o" # set to "o" for scatter plot
for k in range(len(plot_storage)):	
	line, = ax2.plot(plot_storage[k][0], plot_storage[k][2], line_format, color=plot_colors[k], label=plot_legend[k])
	line_handles.append(line)

ax2.legend(handles=line_handles)
ax2.set_title("Std dev of sum vs angle, radius = 0.3 cm")
```

```python
# Plot std dev of mean as function of angle
fig6, ax6 = plt.subplots()

line_handles = []
line_format = "o" # set to "o" for scatter plot
for k in range(len(plot_storage)):	
	line, = ax6.plot(plot_storage[k][0], plot_storage[k][3], line_format, color=plot_colors[k], label=plot_legend[k])
	line_handles.append(line)

ax6.legend(handles=line_handles)
ax6.set_title("Std dev of mean vs angle, radius = 0.3 cm")
```
